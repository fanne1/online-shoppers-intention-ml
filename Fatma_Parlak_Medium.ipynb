# Online Shoppers Intention Veri Seti ile E-Ticaret Satın Alma Davranışının Tahmini

## Giriş

E-ticaretin hızla büyümesiyle birlikte, kullanıcıların bir web sitesindeki davranışlarını anlamak işletmeler için giderek daha önemli hale gelmiştir. Bir ziyaretçinin hangi sayfaları gezdiği, ürün sayfalarında ne kadar zaman geçirdiği, siteden ne kadar hızlı ayrıldığı veya daha önce siteyi ziyaret edip etmediği gibi bilgiler, kullanıcıların satın alma eğilimleri hakkında önemli ipuçları sağlayabilir.

Geleneksel analiz yöntemlerinde bu davranışlar genellikle geçmiş veriler üzerinden incelenirken, makine öğrenmesi sayesinde kullanıcı davranışlarından yararlanılarak gelecekteki bir oturumun satın alma ile sonuçlanıp sonuçlanmayacağı tahmin edilebilir. Bu yaklaşım, özellikle e-ticaret platformlarında potansiyel müşterilerin belirlenmesi, kişiselleştirilmiş kampanyaların oluşturulması ve pazarlama stratejilerinin geliştirilmesi açısından önemli bir kullanım alanına sahiptir.

Bu çalışmada, **Online Shoppers Intention** veri seti kullanılarak web sitesini ziyaret eden kullanıcıların oturum sonunda satın alma gerçekleştirip gerçekleştirmeyeceği tahmin edilmeye çalışılmıştır.

Çalışmanın temel amacı yalnızca yüksek bir doğruluk oranına ulaşmak değildir. Aynı zamanda kullanıcı davranışlarını daha iyi anlamak, satın alma davranışıyla ilişkili olabilecek değişkenleri incelemek ve farklı makine öğrenmesi algoritmalarının bu problem üzerindeki performanslarını karşılaştırmaktır.

## Veri Seti Hakkında

Çalışmada kullanılan veri seti toplam **12.330 oturum** ve **18 değişkenden** oluşmaktadır. Veri setinde kullanıcıların web sitesi içerisindeki davranışlarını ve oturum özelliklerini temsil eden çeşitli değişkenler bulunmaktadır. Bunlar arasında Administrative, Informational, ProductRelated, ProductRelated_Duration, BounceRates, ExitRates, PageValues, Month, TrafficType ve VisitorType gibi değişkenler yer almaktadır. Hedef değişken ise **Revenue** değişkenidir.

`Revenue` değişkeni, ilgili oturumun satın alma ile sonuçlanıp sonuçlanmadığını göstermektedir. Dolayısıyla bu problem bir **ikili sınıflandırma (binary classification)** problemidir. Modelin temel görevi, yeni bir kullanıcı oturumunun sonunda satın alma gerçekleşip gerçekleşmeyeceğini tahmin etmektir.

Bu problem ilk bakışta basit bir sınıflandırma problemi gibi görünse de e-ticaret açısından önemli bir zorluk bulunmaktadır: Satın alma gerçekleştiren kullanıcıların sayısı, satın alma gerçekleştirmeyen kullanıcılara göre daha azdır. Bu nedenle yalnızca Accuracy değerine bakmak modelin gerçek performansını değerlendirmek için yeterli olmayabilir.

Örneğin bir model, satın alma gerçekleştirmeyen kullanıcıları çok başarılı şekilde tahmin ederken satın alma gerçekleştirecek kullanıcıların önemli bir bölümünü kaçırabilir. Bu nedenle çalışmada Accuracy'nin yanında **Precision, Recall ve F1-score** gibi farklı değerlendirme metrikleri de dikkate alınmıştır.

## Neden Sadece Accuracy Yeterli Değil?

E-ticaret satın alma tahmini gibi problemlerde modelin satın alma gerçekleştirecek kullanıcıları ne kadar iyi yakalayabildiği oldukça önemlidir.

Örneğin modelin Accuracy değeri yüksek olabilir ancak satın alma gerçekleştirecek kullanıcıların önemli bir bölümünü "satın almayacak" olarak sınıflandırıyorsa işletme açısından beklenen fayda sağlanamayabilir.

Bu nedenle özellikle **Recall** metriği önem kazanmaktadır. Recall, gerçek satın alma gerçekleştiren kullanıcıların ne kadarının model tarafından doğru şekilde yakalandığını gösterir.

Diğer taraftan Precision, modelin "satın alma gerçekleşecek" şeklinde tahmin ettiği kullanıcıların ne kadarının gerçekten satın alma gerçekleştirdiğini gösterir.

F1-score ise Precision ve Recall değerlerinin dengeli bir şekilde değerlendirilmesini sağlar.

Bu nedenle çalışmada model performansı yalnızca Accuracy üzerinden değerlendirilmemiştir.

## Model Sonuçlarına Genel Bakış

Modellerin karşılaştırılması sonucunda Random Forest modeli yaklaşık **%90 Accuracy** ve **%74 Precision** ile güçlü bir genel performans göstermiştir. Bununla birlikte satın alma gerçekleştiren kullanıcıları yakalama açısından Logistic Regression modelinin **%75 Recall** ve **%61 F1-score** ile daha başarılı olduğu görülmüştür.

Bu sonuç önemli bir noktayı ortaya koymaktadır: Bir makine öğrenmesi projesinde "en iyi model" yalnızca en yüksek Accuracy değerine sahip model olmak zorunda değildir.

Hangi modelin tercih edilmesi gerektiği, problemin amacına ve işletmenin hangi hatayı daha maliyetli gördüğüne bağlıdır.

Eğer amaç genel olarak doğru sınıflandırma oranını artırmaksa Random Forest daha avantajlı olabilir. Ancak amaç satın alma gerçekleştirecek kullanıcıların mümkün olduğunca büyük bir bölümünü yakalamaksa Recall değeri daha yüksek olan bir model tercih edilebilir.

## Değişken Önemlerinin İncelenmesi

Modelleme sürecinin ardından Random Forest modelinin özellik önemleri de incelenmiştir. Bu analiz sayesinde modelin tahmin yaparken hangi değişkenlerden daha fazla yararlandığı görülmeye çalışılmıştır.

Sonuçlara göre **PageValues**, yaklaşık **%38,33** önem değeriyle açık ara en önemli değişken olarak öne çıkmıştır. PageValues'i sırasıyla ExitRates (%10,06), ProductRelated_Duration (%9,38), ProductRelated (%6,55) ve BounceRates (%5,46) takip etmektedir.

Bu sonuç, kullanıcıların web sitesi içerisindeki ürün ve sayfa etkileşimlerinin satın alma tahmininde önemli bir rol oynadığını göstermektedir.

Ancak burada önemli bir ayrım yapmak gerekir. Feature importance değerleri, bir değişken ile satın alma arasında doğrudan **nedensel bir ilişki** olduğunu göstermez. Bu değerler, Random Forest modelinin tahmin yaparken ilgili özellikten ne ölçüde yararlandığını ifade etmektedir.

```python
# ============================================================
# 1. KÜTÜPHANELER VE VERİ SETİNİN YÜKLENMESİ
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Veri setini oku
data = pd.read_csv("online_shoppers_intention.csv")

# İlk 5 satırı görüntüle
data.head()
```

##**Veri Setinin Genel İncelenmesi**

İlk olarak veri seti okunmuş ve veri setinin yapısı incelenmiştir. Satır ve sütun sayısı, değişken isimleri, veri tipleri ve eksik değerler kontrol edilmiştir.

##**Eksik Değerlerin Kontrolü**

Modelleme öncesinde veri setinde eksik gözlem bulunup bulunmadığı kontrol edilmiştir.

```python
# ============================================================
# 3. EKSİK DEĞER KONTROLÜ
# ============================================================

missing_values = data.isnull().sum()

print("Eksik değer kontrolü")
print("=" * 30)

if missing_values.sum() == 0:
    print("✓ Veri setinde eksik değer bulunmamaktadır.")
else:
    print(missing_values[missing_values > 0])
```

##**Hedef Değişkenin İncelenmesi**

Bu projede hedef değişken `Revenue` değişkenidir. Bu değişken, kullanıcının gerçekleştirdiği oturum sonucunda satın alma yapıp yapmadığını göstermektedir.

Sınıfların dağılımı incelendiğinde veri setinde satın alma gerçekleştirmeyen kullanıcıların çoğunlukta olduğu görülmektedir. Bu nedenle model değerlendirmesinde yalnızca accuracy değeri değil, özellikle satın alma sınıfına ait precision, recall ve F1-score değerleri de dikkate alınacaktır.


```python
# ============================================================
# 5. HEDEF DEĞİŞKENİN DAĞILIMI
# ============================================================

revenue_counts = data["Revenue"].value_counts()
revenue_percent = data["Revenue"].value_counts(normalize=True) * 100

revenue_summary = pd.DataFrame({
    "Oturum Sayısı": revenue_counts,
    "Oran (%)": revenue_percent.round(2)
})

revenue_summary.index = [
    "Satın Alma Yok",
    "Satın Alma Var"
]

print(revenue_summary)
```

```python
#2. Boyutunu kontrol et
print("Satır sayısı:", data.shape[0])
print("Sütun sayısı:", data.shape[1])
```

```python
# ============================================================
# 6. GRAFİK: SATIN ALMA DAĞILIMI
# ============================================================

plt.figure(figsize=(7, 5))

plt.bar(
    ["Satın Alma Yok", "Satın Alma Var"],
    revenue_counts.values
)

plt.title("Oturumların Satın Alma Durumu")
plt.xlabel("Satın Alma Durumu")
plt.ylabel("Oturum Sayısı")

plt.tight_layout()
plt.show()
```

##**Keşifsel Veri Analizi**

Hedef değişkenin dağılımını ve satın alma davranışıyla ilişkili değişkenleri incelemek amacıyla çeşitli grafikler oluşturulmuştur.


```python
# ============================================================
# 7. ÜRÜN SAYFASI SAYISI VE SATIN ALMA
# ============================================================

product_mean = data.groupby("Revenue")["ProductRelated"].mean()

plt.figure(figsize=(8, 5))

plt.bar(
    ["Satın Alma Yok", "Satın Alma Var"],
    product_mean.values
)

plt.title("Satın Alma Durumuna Göre Ortalama Ürün Sayfası Sayısı")
plt.xlabel("Satın Alma Durumu")
plt.ylabel("Ortalama Ürün Sayfası Sayısı")

plt.tight_layout()
plt.show()

print("Ortalama ürün sayfası:")
print(product_mean)
```

```python
# ============================================================
# 8. ÜRÜN SAYFALARINDA GEÇİRİLEN SÜRE
# ============================================================

product_duration_mean = (
    data.groupby("Revenue")["ProductRelated_Duration"].mean()
)

plt.figure(figsize=(8, 5))

plt.bar(
    ["Satın Alma Yok", "Satın Alma Var"],
    product_duration_mean.values
)

plt.title(
    "Satın Alma Durumuna Göre "
    "Ürün Sayfalarında Geçirilen Ortalama Süre"
)

plt.xlabel("Satın Alma Durumu")
plt.ylabel("Ortalama Süre")

plt.tight_layout()
plt.show()

print("Ortalama ürün sayfalarında geçirilen süre:")
print(product_duration_mean)
```

```python
# ============================================================
# 9. BOUNCE RATE ANALİZİ
# ============================================================

bounce_mean = data.groupby("Revenue")["BounceRates"].mean()

plt.figure(figsize=(8, 5))

plt.bar(
    ["Satın Alma Yok", "Satın Alma Var"],
    bounce_mean.values
)

plt.title("Satın Alma Durumuna Göre Ortalama Bounce Rate")
plt.xlabel("Satın Alma Durumu")
plt.ylabel("Ortalama Bounce Rate")

plt.tight_layout()
plt.show()

print("Bounce Rate ortalamaları:")
print(bounce_mean)
```

```python
# ============================================================
# 10. EXIT RATE ANALİZİ
# ============================================================

exit_mean = data.groupby("Revenue")["ExitRates"].mean()

print("Exit Rate ortalamaları:")
print(exit_mean)

plt.figure(figsize=(8, 5))

plt.bar(
    ["Satın Alma Yok", "Satın Alma Var"],
    exit_mean.values
)

plt.title("Satın Alma Durumuna Göre Ortalama Exit Rate")
plt.xlabel("Satın Alma Durumu")
plt.ylabel("Ortalama Exit Rate")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 11. PAGE VALUES ANALİZİ
# ============================================================

page_value_mean = data.groupby("Revenue")["PageValues"].mean()

print("PageValues ortalamaları:")
print(page_value_mean)

plt.figure(figsize=(8, 5))

plt.bar(
    ["Satın Alma Yok", "Satın Alma Var"],
    page_value_mean.values
)

plt.title("Satın Alma Durumuna Göre Ortalama PageValues")
plt.xlabel("Satın Alma Durumu")
plt.ylabel("Ortalama PageValues")

plt.tight_layout()
plt.show()
```

##**Aylara Göre Satın Alma Davranışı**

Kullanıcıların ziyaret gerçekleştirdiği aylara göre satın alma oranları karşılaştırılmıştır. Amaç, ziyaret döneminin satın alma davranışı üzerinde farklılık oluşturup oluşturmadığını gözlemlemektir.


```python
# ============================================================
# 12. AYLARA GÖRE SATIN ALMA ORANI
# ============================================================

month_revenue = pd.crosstab(
    data["Month"],
    data["Revenue"],
    normalize="index"
) * 100

month_revenue.columns = [
    "Satın Alma Yok",
    "Satın Alma Var"
]

print(month_revenue.round(2))
```

```python
# ============================================================
# 13. GRAFİK: AYLARA GÖRE SATIN ALMA ORANI
# ============================================================

month_revenue.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Aylara Göre Satın Alma Oranı")
plt.xlabel("Ay")
plt.ylabel("Oran (%)")
plt.xticks(rotation=45)
plt.legend(title="Satın Alma")

plt.tight_layout()
plt.show()
```

##**Ziyaretçi Türüne Göre Satın Alma Davranışı**

Yeni ziyaretçiler ile geri dönen ziyaretçilerin satın alma oranları karşılaştırılmıştır. Böylece ziyaretçi türünün satın alma davranışıyla ilişkisi incelenmiştir.


```python
# ============================================================
# 14. ZİYARETÇİ TİPİNE GÖRE SATIN ALMA
# ============================================================

visitor_revenue = pd.crosstab(
    data["VisitorType"],
    data["Revenue"],
    normalize="index"
) * 100

visitor_revenue.columns = [
    "Satın Alma Yok",
    "Satın Alma Var"
]

print(visitor_revenue.round(2))
```

```python
# ============================================================
# 15. GRAFİK: ZİYARETÇİ TİPİ
# ============================================================

visitor_revenue.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Ziyaretçi Türüne Göre Satın Alma Oranı")
plt.xlabel("Ziyaretçi Türü")
plt.ylabel("Oran (%)")
plt.xticks(rotation=0)
plt.legend(title="Satın Alma")

plt.tight_layout()
plt.show()
```

##**Hafta Sonu ve Hafta İçi Karşılaştırması**

Hafta içi ve hafta sonu gerçekleşen oturumların satın alma oranları karşılaştırılmıştır. Bu analiz, ziyaret gününün satın alma davranışıyla ilişkili olup olmadığını değerlendirmek amacıyla yapılmıştır.


```python
# ============================================================
# 16. HAFTA SONU VE SATIN ALMA İLİŞKİSİ
# ============================================================

weekend_revenue = pd.crosstab(
    data["Weekend"],
    data["Revenue"],
    normalize="index"
) * 100

weekend_revenue.columns = [
    "Satın Alma Yok",
    "Satın Alma Var"
]

weekend_revenue.index = [
    "Hafta İçi",
    "Hafta Sonu"
]

print(weekend_revenue.round(2))
```

```python
# ============================================================
# 17. GRAFİK: HAFTA SONU VE SATIN ALMA
# ============================================================

weekend_revenue.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Hafta Sonu Durumuna Göre Satın Alma Oranı")
plt.xlabel("Ziyaret Zamanı")
plt.ylabel("Oran (%)")
plt.xticks(rotation=0)
plt.legend(title="Satın Alma")

plt.tight_layout()
plt.show()
```

- Hafta sonu ziyaretlerinde satın alma oranı %17,40, hafta içi ziyaretlerinde ise %14,89'dur.

Bu bulgular, kullanıcı davranışlarının satın alma tahmininde kullanılabilecek anlamlı bilgiler içerdiğini göstermektedir.

```python
# ============================================================
# 18. KATEGORİK DEĞİŞKENLER
# ============================================================

categorical_columns = data.select_dtypes(
    include=["object"]
).columns

print("Kategorik değişkenler:")
print(categorical_columns.tolist())
```

```python
# ============================================================
# 19. KATEGORİK DEĞİŞKENLERİN DÖNÜŞTÜRÜLMESİ
# ============================================================

data_encoded = pd.get_dummies(
    data,
    columns=["Month", "VisitorType"],
    drop_first=True,
    dtype=int
)

print("Encoding sonrası veri boyutu:", data_encoded.shape)

data_encoded.head()
```

```python
# ============================================================
# 20. BAĞIMSIZ VE BAĞIMLI DEĞİŞKENLERİN AYRILMASI
# ============================================================

X = data_encoded.drop("Revenue", axis=1)
y = data_encoded["Revenue"]

print("X boyutu:", X.shape)
print("y boyutu:", y.shape)
```

```python
# ============================================================
# 21. EĞİTİM VE TEST VERİLERİNİN AYRILMASI
# ============================================================

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
# ============================================================
# 22. VERİLERİN ÖLÇEKLENMESİ
# ============================================================

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Ölçeklenmiş X_train:", X_train_scaled.shape)
print("Ölçeklenmiş X_test :", X_test_scaled.shape)
```

```python
# ============================================================
# 23. KNN - FARKLI K DEĞERLERİNİN KARŞILAŞTIRILMASI
# ============================================================

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

k_values = [3, 5, 7, 9, 11, 15, 21]

knn_results = []

for k in k_values:

    knn = KNeighborsClassifier(n_neighbors=k)

    knn.fit(X_train_scaled, y_train)

    y_pred_k = knn.predict(X_test_scaled)

    accuracy = accuracy_score(y_test, y_pred_k)
    precision = precision_score(y_test, y_pred_k)
    recall = recall_score(y_test, y_pred_k)
    f1 = f1_score(y_test, y_pred_k)

    knn_results.append([
        k,
        accuracy,
        precision,
        recall,
        f1
    ])

knn_results_df = pd.DataFrame(
    knn_results,
    columns=[
        "K",
        "Accuracy",
        "Precision",
        "Recall",
        "F1"
    ]
)

knn_results_df.round(3)
```

```python
# ============================================================
# 24. KNN PERFORMANS KARŞILAŞTIRMASI
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    knn_results_df["K"],
    knn_results_df["Accuracy"],
    marker="o",
    label="Accuracy"
)

plt.plot(
    knn_results_df["K"],
    knn_results_df["Precision"],
    marker="o",
    label="Precision"
)

plt.plot(
    knn_results_df["K"],
    knn_results_df["Recall"],
    marker="o",
    label="Recall"
)

plt.plot(
    knn_results_df["K"],
    knn_results_df["F1"],
    marker="o",
    label="F1"
)

plt.xlabel("K Değeri")
plt.ylabel("Skor")
plt.title("KNN Modelinde K Değerlerinin Karşılaştırılması")

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 25. FINAL KNN MODELİ
# ============================================================

knn_final = KNeighborsClassifier(n_neighbors=3)

knn_final.fit(
    X_train_scaled,
    y_train
)

y_pred_knn = knn_final.predict(X_test_scaled)

print("=== FINAL KNN MODELİ ===")

print(
    classification_report(
        y_test,
        y_pred_knn,
        target_names=[
            "Satın Alma Yok",
            "Satın Alma Var"
        ]
    )
)
```

```python
# ============================================================
# 26. KNN CONFUSION MATRIX
# ============================================================

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm_knn = confusion_matrix(
    y_test,
    y_pred_knn
)

print("KNN Confusion Matrix:")
print(cm_knn)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_knn,
    display_labels=[
        "Satın Alma Yok",
        "Satın Alma Var"
    ]
)

disp.plot()

plt.title("KNN - Confusion Matrix (K=3)")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 27. LOGISTIC REGRESSION
# ============================================================

from sklearn.linear_model import LogisticRegression

logistic = LogisticRegression(
    max_iter=1000,
    random_state=42
)

logistic.fit(
    X_train_scaled,
    y_train
)

y_pred_logistic = logistic.predict(
    X_test_scaled
)

print("=== LOGISTIC REGRESSION ===")

print(
    classification_report(
        y_test,
        y_pred_logistic,
        target_names=[
            "Satın Alma Yok",
            "Satın Alma Var"
        ]
    )
)
```

```python
# ============================================================
# 28. DECISION TREE
# ============================================================

from sklearn.tree import DecisionTreeClassifier

tree = DecisionTreeClassifier(
    class_weight="balanced",
    random_state=42
)

tree.fit(
    X_train,
    y_train
)

y_pred_tree = tree.predict(
    X_test
)

print("=== DECISION TREE ===")

print(
    classification_report(
        y_test,
        y_pred_tree,
        target_names=[
            "Satın Alma Yok",
            "Satın Alma Var"
        ]
    )
)
```

```python
# ============================================================
# 29. RANDOM FOREST
# ============================================================

from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42
)

rf.fit(
    X_train,
    y_train
)

y_pred_rf = rf.predict(
    X_test
)

print("=== RANDOM FOREST ===")

print(
    classification_report(
        y_test,
        y_pred_rf,
        target_names=[
            "Satın Alma Yok",
            "Satın Alma Var"
        ]
    )
)
```

```python
# ============================================================
# 30. RANDOM FOREST CONFUSION MATRIX
# ============================================================

cm_rf = confusion_matrix(
    y_test,
    y_pred_rf
)

print("Random Forest Confusion Matrix:")
print(cm_rf)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_rf,
    display_labels=[
        "Satın Alma Yok",
        "Satın Alma Var"
    ]
)

disp.plot()

plt.title("Random Forest - Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 31. MODEL KARŞILAŞTIRMASI
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

model_results = []

models = {
    "KNN (K=3)": y_pred_knn,
    "Logistic Regression": y_pred_logistic,
    "Decision Tree": y_pred_tree,
    "Random Forest": y_pred_rf
}

for model_name, predictions in models.items():

    model_results.append([
        model_name,
        accuracy_score(y_test, predictions),
        precision_score(y_test, predictions),
        recall_score(y_test, predictions),
        f1_score(y_test, predictions)
    ])

model_results_df = pd.DataFrame(
    model_results,
    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1"
    ]
)

model_results_df.set_index("Model").round(3)
```

```python
# ============================================================
# 32. RANDOM FOREST FEATURE IMPORTANCE
# ============================================================

feature_importance = pd.Series(
    rf.feature_importances_,
    index=X_train.columns
).sort_values(ascending=False)

print("Özellik önem sıralaması:")
print(feature_importance)
```

```python
# ============================================================
# 33. EN ÖNEMLİ 10 ÖZELLİK
# ============================================================

top_features = (
    feature_importance
    .head(10)
    .sort_values()
)

plt.figure(figsize=(10, 6))

top_features.plot(
    kind="barh"
)

plt.title(
    "Random Forest - En Önemli 10 Değişken"
)

plt.xlabel("Önem Düzeyi")
plt.ylabel("Değişken")

plt.tight_layout()
plt.show()
```
